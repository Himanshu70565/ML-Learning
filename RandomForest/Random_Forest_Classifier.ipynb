```python
import numpy as np;
import pandas as pd;
```

```python
df = pd.read_csv('../datasets/heart.csv')
df.head(5)
```

```python
df.shape
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y)
```

# Checking accuracy on the above dataset using RandomForest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
clf = RandomForestClassifier(n_estimators=100)
clf.fit(X_train, y_train)
y_test_predicted = clf.predict(X_test)
accuracy_score(y_test, y_test_predicted)
```

```python
# Implementing cross_val_score
from sklearn.model_selection import cross_val_score
np.mean(cross_val_score(RandomForestClassifier(n_estimators=100), X, y))
```

# Now checking the same accuracy on the same dataset using Decision Tree Classifier

```python
from sklearn.tree import DecisionTreeClassifier
dec_clf = DecisionTreeClassifier()
dec_clf.fit(X_train, y_train)
y_test_predicted = dec_clf.predict(X_test);
accuracy_score(y_test, y_test_predicted)
```

```python
np.mean(cross_val_score(DecisionTreeClassifier(), X, y))
```
