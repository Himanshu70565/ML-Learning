```python
import pandas as pd;
import numpy as np;
from sklearn.cluster import KMeans;
```

```python
df = pd.read_csv("../datasets/student_clustering.csv");
```

```python
df
```

```python
import seaborn as sns;
sns.scatterplot(df, x = 'cgpa', y = 'iq');
```

```python
wcss_list = [];

for i in range(1, 11):
    kmeans = KMeans(n_clusters = i)
    labels = kmeans.fit_predict(df)
    wcss_list.append(kmeans.inertia_);
```

```python
import matplotlib.pyplot as plt;
```

```python
# Printing the Elbow Plot
plt.plot(np.arange(1,11), wcss_list);
```

```python
# From above, we can see that the elbow is created at n_cluster = 4
```

```python
kmeans = KMeans(n_clusters = 4);
labels = kmeans.fit_predict(df);
labels
```
