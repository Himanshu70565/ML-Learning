```python
# Read, inspect, and transform Titanic data, then write it back out

# df = pd.read_csv("")
# method to read the titanic.csv file
df = pd.read_csv("./titanic.csv")
# there are other method similair to read csv like read_excel, read_clipbaord that we can explore

# In case suppose you want to read only tope 12 rows 
df.head(n = 12)
df.tail(n = 12)


# This will add prefix = Himanshu_ and it will add in column heading
df.add_prefix(prefix = "Himanshu_", axis = 1).head(12)

# df.count() -> This will give non-NA cells for each column or row, for column set axis = 0 else row wise axis = 1



# Getting count of the data element which are not-nan
print(df["Age"].count())

# Getting first 5 rows of the data frame
print(df["Age"].head(5))
# Implementing iterator over the dataframe - series
df["Age"] = df["Age"].map(lambda x: x + 1)
print(df["Age"].head(5))

# Similiar to read_csv we also have to_csv, to_excel ...
df.to_csv("titanic_2.csv")
```
