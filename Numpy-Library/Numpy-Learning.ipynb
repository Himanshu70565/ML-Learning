```python
# Create an int16 array to show explicit dtype handling
b = np.array([1,2,3], np.int16)
# b[0] = 1231212321 - Here if we try to assign a big value that does not fit into the range then it will give error overflow
```

```python
# How to perform slicing operation on numpay array
a[3:]
a[:3]
# Both of the above mentioned give different output since they both are serving different purpose
```

```python
# Concept - In case we are performing slicing operation on numpay array then it returns a view of th array, it does not give a copy of the list
# Step 1 : Let's do this witn Numpy
b = a[:3]
print(a,b)
# print(type(a), type(b))

b[0] = 212
print(a,b)
print(b.base)


# Step 2: Let's do this with Python
a_pyton = [1,2,3,4,5,6]
b_python = a[:3]
print(a_pyton, b_python);
b_python[0]=212
print(a_pyton, b_python);
```

```python
# How to determine if the return of numpy function is a view or a copy of the original array

array = np.array([1,2,3])
array

array_view = array[:1]
array_copy = array.copy()
print(array_view.base, array_copy.base)
# Here we get None, it shows that it is a copy of the original array
# These type of functionality are not present in python, but numpy library supports this.
```

```python
# Function to check whether the data is iterable or not
def is_iterable(obj):
    try:
        iter(obj)
        return True
    except TypeError:
        return False

is_iterable([2])
```

```python
# Concept: How to create array from existing data
a1 = np.array([[1, 1],
               [2, 2]])

a2 = np.array([[3, 3],
               [4, 4]])

vstack_array = np.concatenate((a1,a2), axis = 0)
hstack_array = np.concatenate((a1,a2), axis = 1)

print(vstack_array)
print(hstack_array)

hstack = np.hstack((a1,a2))
vstack = np.vstack((a1,a2))

print(vstack)
print(hstack)
```

```python
# Basic Array Operation
matrix1 = np.array([[1,2],[3,4]])
matrix2 = np.array([[5,6],[7,8]])

# Sum of two matrix
sum_matrix = np.add(matrix1, matrix2)
print(sum_matrix)

# Addition of two matrix in python
matrix3 = matrix1 + matrix2
print(matrix3)

# Here for both above operation result is same, but one use Python loop for calcualtion and other use direct c compiled code for caluclation.
# Due to this reason numpy is faster than python add operation and that's why all other library use the concept of numpy for performing matrix operation


# Subtraction of two matrix
matrix4 = np.subtract(matrix3, matrix2)
print(matrix4.base)

matrix5 = matrix3 - matrix2;
# print(matrix5.base) - None - This simply shows that there is copy returnd over ehre
```
