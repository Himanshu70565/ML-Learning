```python
# Import Matplotlib and NumPy for plotting and array creation
import matplotlib.pyplot as plt;
import numpy as np;
```

```python
# Create sample x and y values for a sine wave plot
x = np.linspace(0, 10, 200);
y = np.sin(x);
```

```python
# Plot the sine wave using the object-oriented Matplotlib API
fig, ax = plt.subplots();
ax.plot(x, y)
plt.show()
```

```python
# Create another simple data set with matching x and y values
x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])
```

```python
# Plot the linear data set to compare with the sine wave example
fig, ax = plt.subplots();
ax.plot(x, y)
plt.show()
```

```python
# Build x and y values for a labeled sine curve
x = np.linspace(0, 5, 11)
y = np.sin(x)
```

```python
# Draw a red line and add axis labels plus a title
plt.plot(x,y,'r-')
plt.xlabel("X Label Age")
plt.ylabel("Y Lable")
plt.title("Title Label")
```

```python
# Compare two subplots side by side with scaled sine curves
plt.subplot(1,2,1)
plt.plot(x,y,'r-')

plt.subplot(1,2,2)
plt.plot(2*x,2*y,'b-')

plt.plot(3*x,3*y,'b-')
```

```python
# Object Oriented Way for Matplotlib
fig = plt.figure();
axes_1 = fig.add_axes([0.5,0.5,0.5,0.5])
axes_1.plot(x, y)


axes_2 = fig.add_axes([1,1,1,1])
axes_2.plot(2*x, 2*y)
```

```python
# Creating a Figure object
fig = plt.figure();

# Fix edge color
fig.patch.set_edgecolor('b')
fig.patch.set_linewidth(5)


axes1 = fig.add_axes([0, 0, 0.5, 1]);
axes1.set_xlabel("Axes1 - X Label");
axes1.set_ylabel("Axes1 - Y Label");
axes1.set_title("Left Plot")
axes1.plot(x,y);

axes2 = fig.add_axes([0.7, 0, 0.5, 1]);
axes2.set_xlabel("Axes2 - X Label");
axes2.set_ylabel("Axes2 - Y Label");
axes2.set_title("Right Plot")
axes2.plot(y,x);
```

```python
fig, axes_array = plt.subplots(nrows = 3, ncols=3);

iteration_count = 0;


for row, axes_row in enumerate(axes_array):
    for col, axes in enumerate(axes_row):
        iteration_count = iteration_count + 1;
        axes_array[row][col].plot(x,y)
        axes_array[row][col].set_xlabel(f"X Label: {iteration_count}");
        axes_array[row][col].set_ylabel(f"Y Label: {iteration_count}");
        axes.set_title(f"Title: {iteration_count}");        

plt.tight_layout();
```

```python
fig, axes = plt.subplots(ncols=2)
plt.tight_layout();

for index, current_axes in enumerate(axes):
    current_axes.plot(x,y)
    current_axes.set_xlabel(f"X Label: {index}")
    current_axes.set_ylabel(f"Y Label: {index}")
```

```python
fig, axes = plt.subplots(nrows = 2, ncols = 1, figsize=[8,5]);

# Fix edge color
fig.patch.set_edgecolor('b')
fig.patch.set_linewidth(5)

axes[0].plot(x,y);
axes[1].plot(np.sin(10000*x),y);

plt.tight_layout()
# axes.set_title("Title Page")
```

```python
# Plot multiple lines on one axis and add a legend
fig = plt.figure();

axes = fig.add_axes([0,0,1,1])
axes.plot(x,x,label='Number');
axes.plot(x,2*x, label='Number Square');
axes.plot(x,3*x, label='Number Triple');

axes.legend()
```

```python
# Combine line styles, transparency, and legend placement
fig, axes = plt.subplots()
axes = fig.add_axes([0,0,1,1])
axes.plot(x,x,'o:r', label='Number', color='k', alpha = 0.5);
axes.plot(x,2*x, label='Number Square');
axes.plot(x,3*x, label='Number Triple');

axes.legend(loc=10)
```
