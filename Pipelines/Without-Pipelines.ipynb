```python
import pandas as pd;
import numpy as np;
from sklearn.model_selection import train_test_split;
from sklearn.impute import SimpleImputer;
from sklearn.preprocessing import OneHotEncoder;
```

```python
# Importing the dataframe
df = pd.read_csv('../datasets/train.csv');
```

```python
df
```

```python
df.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'],inplace=True, axis = 1)
```

```python
df.head(5)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df['Survived'], test_size=0.3, random_state=0)
```

```python
df.isnull().sum()
```

```python
# Applying Imputation on the "Age" and "Embarked" column since they contain missing values
si_age = SimpleImputer(strategy='mean');
si_transform = SimpleImputer(strategy='most_frequent');

X_train_age = si_age.fit_transform(X_train[["Age"]])
X_train_embarked = si_transform.fit_transform(X_train[["Embarked"]])

X_test_age = si_age.transform(X_test[["Age"]])
X_test_embarked = si_transform.transform(X_test[["Embarked"]])
```

```python
# Performing One Hot Encoder on the "Embarked" and "Sex" column
ohe_embarked = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore');
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked);
X_test_embarked = ohe_embarked.transform(X_test_embarked);

ohe_gender = OneHotEncoder(drop='first', sparse_output=False);
X_train_gender = ohe_gender.fit_transform(X_train[['Sex']]);
X_test_gender = ohe_gender.transform(X_test[['Sex']]);
```

```python
df1 = pd.concat([pd.DataFrame(X_train_age),
     pd.DataFrame(X_train_embarked),
     pd.DataFrame(X_train_gender)],
    axis=1
).reset_index(drop=True)

X_train_remaining = X_train.drop(['Sex', 'Age', 'Embarked'], axis = 1).reset_index(drop=True)
X_train_final = pd.concat([X_train_remaining, df1], axis = 1)
X_train_final
```
